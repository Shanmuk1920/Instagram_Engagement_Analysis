```python
import pandas as pd
import os

folder = "/content/instagram_dataset"

# Load all CSV files
likes = pd.read_csv(f"{folder}/likes.csv")
tags = pd.read_csv(f"{folder}/tags.csv")
photo_tags = pd.read_csv(f"{folder}/photo_tags.csv")
follows = pd.read_csv(f"{folder}/follows.csv")
photos = pd.read_csv(f"{folder}/photos.csv")
users = pd.read_csv(f"{folder}/users.csv")
comments = pd.read_csv(f"{folder}/comments.csv")

# Store datasets in a dictionary
datasets = {
    "Likes": likes,
    "Tags": tags,
    "Photo Tags": photo_tags,
    "Follows": follows,
    "Photos": photos,
    "Users": users,
    "Comments": comments
}

# Display dataset information
for name, df in datasets.items():
    print(f"\n{'='*45}")
    print(f"📊 {name.upper()}")
    print(f"{'='*45}")

    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())

    print("\nFirst 3 rows:")
    display(df.head(3))

    print("\nMissing Values:")
    print(df.isnull().sum())
```

```python
import pandas as pd

# 1. Standardize column names
for name, data in datasets.items():
    data.columns = (
        data.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
    )

# 2. Rename important columns
photos.rename(columns={
    "user_id": "owner_id",
    "created_dat": "post_datetime"
}, inplace=True)

comments.rename(columns={
    "photo_id": "photo",
    "created_timestamp": "comment_datetime"
}, inplace=True)

likes.rename(columns={
    "created_time": "like_datetime"
}, inplace=True)

follows.rename(columns={
    "created_time": "follow_datetime"
}, inplace=True)

# 3. Convert dates
photos["post_datetime"] = pd.to_datetime(
    photos["post_datetime"], dayfirst=True, errors="coerce"
)

comments["comment_datetime"] = pd.to_datetime(
    comments["comment_datetime"], dayfirst=True, errors="coerce"
)

likes["like_datetime"] = pd.to_datetime(
    likes["like_datetime"], dayfirst=True, errors="coerce"
)

follows["follow_datetime"] = pd.to_datetime(
    follows["follow_datetime"], dayfirst=True, errors="coerce"
)

# 4. Remove duplicate rows
for name, data in datasets.items():
    data.drop_duplicates(inplace=True)

# 5. Check cleaned datasets
print("✅ Data Cleaning Completed!\n")

for name, data in datasets.items():
    print(f"{name}: {data.shape}")

print("\nMissing dates after conversion:")
print("Photos:", photos["post_datetime"].isna().sum())
print("Comments:", comments["comment_datetime"].isna().sum())
print("Likes:", likes["like_datetime"].isna().sum())
print("Follows:", follows["follow_datetime"].isna().sum())

print("\nCleaned Photos Dataset:")
display(photos.head())
```

```python
# STEP 6: INSTAGRAM ENGAGEMENT ANALYSIS

# 1. Count likes for each photo
likes_count = (
    likes.groupby("photo")
    .size()
    .reset_index(name="total_likes")
)

# 2. Count comments for each photo
comments_count = (
    comments.groupby("photo")
    .size()
    .reset_index(name="total_comments")
)

# 3. Count followers for each user
followers_count = (
    follows.groupby("followee")
    .size()
    .reset_index(name="total_followers")
)

# 4. Merge likes with photos
engagement = photos.merge(
    likes_count,
    left_on="id",
    right_on="photo",
    how="left"
)

engagement.drop(columns=["photo"], inplace=True)

# 5. Merge comments
engagement = engagement.merge(
    comments_count,
    left_on="id",
    right_on="photo",
    how="left"
)

engagement.drop(columns=["photo"], inplace=True)

# 6. Merge followers
engagement = engagement.merge(
    followers_count,
    left_on="owner_id",
    right_on="followee",
    how="left"
)

engagement.drop(columns=["followee"], inplace=True)

# 7. Replace missing counts with zero
engagement[
    ["total_likes", "total_comments", "total_followers"]
] = engagement[
    ["total_likes", "total_comments", "total_followers"]
].fillna(0)

# 8. Calculate total engagement
engagement["total_engagement"] = (
    engagement["total_likes"] +
    engagement["total_comments"]
)

# 9. Calculate engagement rate (%)
engagement["engagement_rate"] = (
    engagement["total_engagement"] /
    engagement["total_followers"].replace(0, float("nan"))
) * 100

# 10. Display results
print("✅ Engagement Metrics Calculated Successfully!")

display(
    engagement[
        [
            "id",
            "owner_id",
            "total_likes",
            "total_comments",
            "total_followers",
            "total_engagement",
            "engagement_rate"
        ]
    ].head(10)
)

print("\nAverage Engagement Rate:",
      round(engagement["engagement_rate"].mean(), 2), "%")

print("\nTotal Likes:", engagement["total_likes"].sum())
print("Total Comments:", engagement["total_comments"].sum())
```

```python
# STEP 7: BEST POSTING DAYS AND TIMES

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Extract posting day and hour
engagement["posting_day"] = engagement["post_datetime"].dt.day_name()
engagement["posting_hour"] = engagement["post_datetime"].dt.hour

day_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

# 2. Average engagement by day
day_engagement = (
    engagement.groupby("posting_day")["engagement_rate"]
    .mean()
    .reindex(day_order)
    .dropna()
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=day_engagement.index,
    y=day_engagement.values,
    color="skyblue"
)

plt.title("Average Instagram Engagement Rate by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Average Engagement Rate (%)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# 3. Average engagement by posting hour
hour_engagement = (
    engagement.groupby("posting_hour")["engagement_rate"]
    .mean()
)

plt.figure(figsize=(10, 5))

sns.lineplot(
    x=hour_engagement.index,
    y=hour_engagement.values,
    marker="o"
)

plt.title("Average Instagram Engagement Rate by Posting Hour")
plt.xlabel("Posting Hour (24-Hour Format)")
plt.ylabel("Average Engagement Rate (%)")
plt.xticks(range(24))
plt.tight_layout()
plt.show()

# 4. Day and hour heatmap
heatmap_data = engagement.pivot_table(
    index="posting_day",
    columns="posting_hour",
    values="engagement_rate",
    aggfunc="mean"
)

heatmap_data = heatmap_data.reindex(day_order).dropna(how="all")

plt.figure(figsize=(12, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu"
)

plt.title("Instagram Engagement Heatmap: Day vs Hour")
plt.xlabel("Posting Hour")
plt.ylabel("Posting Day")
plt.tight_layout()
plt.show()

# 5. Display posting-time coverage
print("\nPosting Days Available:")
print(engagement["posting_day"].value_counts())

print("\nPosting Hours Available:")
print(engagement["posting_hour"].value_counts().sort_index())

# 6. Identify best posting day and hour
if not day_engagement.empty:
    print("\nBest Posting Day:", day_engagement.idxmax())

if not hour_engagement.empty:
    print("Best Posting Hour:", hour_engagement.idxmax())

print("\nUnique Posting Timestamps:",
      engagement["post_datetime"].nunique())
```

```python
# STEP 8: HASHTAG PERFORMANCE ANALYSIS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Merge photo tags with hashtag names
hashtag_data = photo_tags.merge(
    tags,
    left_on="tag_id",
    right_on="id",
    how="left"
)

# 2. Merge hashtag data with engagement metrics
hashtag_engagement = hashtag_data.merge(
    engagement[
        ["id", "total_likes", "total_comments",
         "total_engagement", "engagement_rate"]
    ],
    left_on="photo",
    right_on="id",
    how="left"
)

# 3. Calculate hashtag performance
hashtag_performance = (
    hashtag_engagement.groupby("tag_text")
    .agg(
        total_uses=("photo", "nunique"),
        avg_likes=("total_likes", "mean"),
        avg_comments=("total_comments", "mean"),
        avg_engagement=("total_engagement", "mean"),
        avg_engagement_rate=("engagement_rate", "mean")
    )
    .reset_index()
)

# 4. Sort hashtags by average engagement
hashtag_performance = hashtag_performance.sort_values(
    by="avg_engagement",
    ascending=False
)

print("✅ Hashtag Performance Analysis Completed!")

display(hashtag_performance.head(10))

# 5. Top 10 most frequently used hashtags
top_hashtags = hashtag_performance.nlargest(
    10, "total_uses"
)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=top_hashtags,
    x="total_uses",
    y="tag_text",
    color="skyblue"
)

plt.title("Top 10 Most Frequently Used Instagram Hashtags")
plt.xlabel("Number of Posts")
plt.ylabel("Hashtag")
plt.tight_layout()
plt.show()

# 6. Top 10 hashtags by average engagement
top_engagement_hashtags = hashtag_performance.head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=top_engagement_hashtags,
    x="avg_engagement",
    y="tag_text",
    color="lightgreen"
)

plt.title("Top 10 Hashtags by Average Engagement")
plt.xlabel("Average Likes + Comments per Post")
plt.ylabel("Hashtag")
plt.tight_layout()
plt.show()

# 7. Count hashtags per post
hashtag_counts = (
    photo_tags.groupby("photo")["tag_id"]
    .nunique()
    .reset_index(name="hashtag_count")
)

hashtag_analysis = engagement.merge(
    hashtag_counts,
    left_on="id",
    right_on="photo",
    how="left"
)

hashtag_analysis["hashtag_count"] = (
    hashtag_analysis["hashtag_count"].fillna(0)
)

# 8. Analyze engagement by hashtag count
hashtag_count_performance = (
    hashtag_analysis.groupby("hashtag_count")
    .agg(
        post_count=("id", "count"),
        avg_engagement=("total_engagement", "mean")
    )
    .reset_index()
)

print("\nEngagement by Hashtag Count:")
display(hashtag_count_performance)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=hashtag_count_performance,
    x="hashtag_count",
    y="avg_engagement",
    color="orange"
)

plt.title("Average Instagram Engagement by Number of Hashtags")
plt.xlabel("Number of Hashtags")
plt.ylabel("Average Likes + Comments")
plt.tight_layout()
plt.show()
```

```python
# STEP 9: INSTAGRAM CONTENT TYPE ANALYSIS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Standardize content type and filter values
engagement["photo_type"] = (
    engagement["photo_type"]
    .astype(str)
    .str.strip()
    .str.lower()
)

engagement["insta_filter_used"] = (
    engagement["insta_filter_used"]
    .astype(str)
    .str.strip()
    .str.lower()
)

# 2. Count posts by content type
print("Instagram Content Types:")
print(engagement["photo_type"].value_counts())

# 3. Calculate content type performance
content_performance = (
    engagement.groupby("photo_type")
    .agg(
        total_posts=("id", "count"),
        avg_likes=("total_likes", "mean"),
        avg_comments=("total_comments", "mean"),
        avg_engagement=("total_engagement", "mean"),
        avg_engagement_rate=("engagement_rate", "mean")
    )
    .reset_index()
    .sort_values("avg_engagement", ascending=False)
)

print("\n✅ Content Type Performance:")
display(content_performance.round(2))

# 4. Plot average engagement by content type
plt.figure(figsize=(9, 5))

sns.barplot(
    data=content_performance,
    x="photo_type",
    y="avg_engagement",
    color="skyblue"
)

plt.title("Average Instagram Engagement by Content Type")
plt.xlabel("Content Type")
plt.ylabel("Average Likes + Comments")
plt.tight_layout()
plt.show()

# 5. Analyze Instagram filter usage
filter_performance = (
    engagement.groupby("insta_filter_used")
    .agg(
        total_posts=("id", "count"),
        avg_likes=("total_likes", "mean"),
        avg_comments=("total_comments", "mean"),
        avg_engagement=("total_engagement", "mean"),
        avg_engagement_rate=("engagement_rate", "mean")
    )
    .reset_index()
)

print("\n✅ Instagram Filter Performance:")
display(filter_performance.round(2))

# 6. Plot filter performance
plt.figure(figsize=(8, 5))

sns.barplot(
    data=filter_performance,
    x="insta_filter_used",
    y="avg_engagement",
    color="lightgreen"
)

plt.title("Instagram Engagement: Filter vs No Filter")
plt.xlabel("Instagram Filter Used")
plt.ylabel("Average Likes + Comments")
plt.tight_layout()
plt.show()

# 7. Identify highest-performing content type
if not content_performance.empty:
    print(
        "\nHighest Average Engagement Content Type:",
        content_performance.iloc[0]["photo_type"]
    )
```

```python
# STEP 10: FOLLOWER ACTIVITY AND GROWTH SIGNALS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Count followers for each account
top_followed = (
    follows.groupby("followee")
    .size()
    .reset_index(name="follower_count")
    .sort_values("follower_count", ascending=False)
    .head(10)
)

# Add usernames
top_followed = top_followed.merge(
    users[["id", "name"]],
    left_on="followee",
    right_on="id",
    how="left"
)

print("TOP 10 ACCOUNTS BY FOLLOWER COUNT")
display(top_followed[["name", "follower_count"]])

# 2. Plot top accounts by follower count
plt.figure(figsize=(10, 5))

sns.barplot(
    data=top_followed,
    x="follower_count",
    y="name",
    color="skyblue"
)

plt.title("Top 10 Accounts by Number of Followers")
plt.xlabel("Number of Followers")
plt.ylabel("Instagram Username")
plt.tight_layout()
plt.show()

# 3. Analyze active and inactive followers
activity_summary = (
    follows["is_follower_active"]
    .value_counts()
    .reset_index()
)

activity_summary.columns = ["activity_status", "count"]

print("\nFOLLOWER ACTIVITY SUMMARY")
display(activity_summary)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=activity_summary,
    x="activity_status",
    y="count",
    color="lightgreen"
)

plt.title("Active vs Inactive Followers")
plt.xlabel("Activity Status (1 = Active, 0 = Inactive)")
plt.ylabel("Number of Follow Relationships")
plt.tight_layout()
plt.show()

# 4. Analyze public/private account status
account_status = (
    follows["followee_acc_status"]
    .str.lower()
    .value_counts()
    .reset_index()
)

account_status.columns = ["account_status", "count"]

print("\nACCOUNT STATUS SUMMARY")
display(account_status)

# 5. Analyze verified users
verified_summary = (
    users["verified_status"]
    .str.lower()
    .value_counts()
    .reset_index()
)

verified_summary.columns = ["verified_status", "count"]

print("\nVERIFIED USER SUMMARY")
display(verified_summary)

# 6. Analyze the relationship between post count and followers
user_followers = (
    follows.groupby("followee")
    .size()
    .reset_index(name="follower_count")
)

user_analysis = users.merge(
    user_followers,
    left_on="id",
    right_on="followee",
    how="left"
)

user_analysis["follower_count"] = (
    user_analysis["follower_count"].fillna(0)
)

print("\nPOST COUNT AND FOLLOWER COUNT CORRELATION:")

correlation = user_analysis[
    ["post_count", "follower_count"]
].corr().iloc[0, 1]

print(round(correlation, 3))

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=user_analysis,
    x="post_count",
    y="follower_count"
)

plt.title("Relationship Between Post Count and Followers")
plt.xlabel("Number of Posts")
plt.ylabel("Number of Followers")
plt.tight_layout()
plt.show()
```
